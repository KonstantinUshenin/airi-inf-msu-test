## 2. NumPy

**Ключевое отличие от списков Python:** операции применяются *поэлементно*, без циклов.

```python
import numpy as np

# Создание массива
a = np.array([1, 4, 9, 16, 25])
print("shape:", a.shape)   # атрибут, НЕ метод — без скобок!
print("dtype:", a.dtype)   # тип данных элементов
```

```python
# linspace — равномерная сетка точек
x = np.linspace(0, 2 * np.pi, 100)  # 100 точек от 0 до 2π
print("Первые 5:", x[:5])            # срез [:5] = элементы 0,1,2,3,4
print("Последние 5:", x[-5:])        # срез [-5:] = последние 5
```

```python
# Поэлементные операции (вместо циклов!)
print("a ** 2  =", a ** 2)          # возведение в квадрат
print("sqrt(a) =", np.sqrt(a))      # корень (np.square — это КВАДРАТ, не корень!)

b = np.array([10, 20, 30, 40, 50])
print("a + b   =", a + b)           # поэлементное сложение
```

```python
# Broadcasting — NumPy "растягивает" массив под нужную форму
a = np.arange(5)         # [0, 1, 2, 3, 4] — это NumPy array
print(a + 10)             # скаляр 10 прибавляется к каждому элементу

# ВАЖНО: для обычного списка Python "+" — это конкатенация!
print(list(range(5)) + [10])  # [0, 1, 2, 3, 4, 10] — другое поведение
```

```python
# 2D broadcasting: чтобы сложить вектор-столбец и вектор-строку, нужен reshape
rows = np.arange(3).reshape(-1, 1)  # столбец (3, 1)
cols = np.arange(4)                  # строка  (4,)
print(rows + cols)                    # результат (3, 4): a_ij = i + j
```

## 3. Matplotlib: первый график

Два интерфейса:
- `plt.plot(...)` — быстрый (pyplot)
- `fig, ax = plt.subplots()` → `ax.plot(...)` — для subplots

```python
import matplotlib.pyplot as plt

x = np.linspace(0, 2 * np.pi, 200)  # много точек = гладкая кривая

plt.plot(x, np.sin(x), label="sin(x)")
plt.plot(x, np.cos(x), label="cos(x)")
plt.xlabel("x")
plt.ylabel("y")            # не оставляйте пустым!
plt.title("sin и cos")
plt.legend()                # без label= в plot() — легенда будет пустой
plt.grid(True)

# ВАЖНО: savefig ПЕРЕД show! show() очищает фигуру
plt.savefig("demo_sin_cos.png", dpi=150)
plt.show()
```

## 4. Оформление: стили, цвета, linewidth

```python
x = np.linspace(0, 4 * np.pi, 500)
y = np.exp(-x / 3) * np.sin(2 * x)   # УМНОЖЕНИЕ, не сложение!

plt.figure(figsize=(10, 4))
plt.plot(x, y, color="steelblue", linewidth=2, linestyle="-")  # linewidth — число, не строка!
plt.title("Затухающие колебания")
plt.grid(True)
plt.show()
```

```python
# Стили — меняют внешний вид всех последующих графиков
print("Доступные стили:", plt.style.available[:5], "...")

# "seaborn" устарел! Нужно "seaborn-v0_8" или другой из списка
plt.style.use("seaborn-v0_8")

plt.figure(figsize=(10, 4))
plt.plot(x, y, linewidth=2)
plt.title("Тот же график в стиле seaborn-v0_8")
plt.show()

plt.style.use("default")  # вернём стиль по умолчанию
```

## 5. Subplots: несколько графиков

```python
fig, axes = plt.subplots(1, 2, figsize=(10, 4))  # 1 строка, 2 столбца

x = np.linspace(-5, 5, 300)

axes[0].plot(x, x ** 2)
axes[0].set_title("$y = x^2$")
axes[0].grid(True)

# sin(x)/x: в нуле будет nan → используем np.sinc(x/pi)
axes[1].plot(x, np.sinc(x / np.pi))  # sinc(t) = sin(πt)/(πt), поэтому sinc(x/π) = sin(x)/x
axes[1].set_title("$y = \\sin(x) / x$")
axes[1].grid(True)

fig.tight_layout()  # чтобы графики не налезали друг на друга
plt.show()
```

## 6. Scatter, bar, stem, colorbar

```python
np.random.seed(42)

fig, axes = plt.subplots(1, 3, figsize=(15, 4))  # ВАЖНО: (1, 3) — 1 строка, 3 столбца

# Scatter: c= задаёт цвет точек, cmap — цветовую карту
x = np.random.randn(200)
y = 0.5 * x + np.random.randn(200) * 0.3
sc = axes[0].scatter(x, y, c=y, cmap="viridis", alpha=0.7)  # c=y — цвет по y
fig.colorbar(sc, ax=axes[0], label="y")
axes[0].set_title("Scatter")

# Bar
k = np.arange(1, 11)
axes[1].bar(k, np.sin(k), color="coral")  # sin(k), НЕ sin(k*pi)!
axes[1].set_title("sin(k)")

# Stem
n = np.arange(1, 21)            # начинаем с 1, чтобы не делить на 0
axes[2].stem(n, (-1)**n / n)
axes[2].set_title("$(-1)^n / n$")

fig.tight_layout()
plt.show()
```

## 7. Гистограммы и density

```python
np.random.seed(42)
data = np.random.randn(10000)

plt.figure(figsize=(8, 4))

# density=True нормирует так, что площадь под гистограммой = 1
# Без density=True ось Y — абсолютные частоты (сотни/тысячи),
# и теоретическая кривая (от 0 до ~0.4) будет незаметна
plt.hist(data, bins=50, density=True, color="skyblue", edgecolor="black", label="Данные")

x = np.linspace(-4, 4, 300)
plt.plot(x, (1/np.sqrt(2*np.pi)) * np.exp(-x**2/2), "r-", lw=2, label="N(0,1) плотность")

plt.legend()
plt.title("density=True уравнивает масштабы")
plt.show()
```

## 8. 2D-визуализация: meshgrid, imshow, contourf

```python
# meshgrid создаёт 2D-сетку из двух 1D-массивов
x = np.arange(-3, 3, 0.05)
y = np.arange(-3, 3, 0.05)   # шаг ОДИНАКОВЫЙ по обеим осям!
X, Y = np.meshgrid(x, y)
Z = np.sin(X**2 + Y**2)

fig, axes = plt.subplots(1, 2, figsize=(12, 5))

# imshow: origin="lower" — чтобы Y шёл снизу вверх (как в математике)
im = axes[0].imshow(Z, extent=[-3,3,-3,3], origin="lower", cmap="viridis")
fig.colorbar(im, ax=axes[0])
axes[0].set_title("imshow (origin='lower'!)")

# contourf + clabel
cf = axes[1].contourf(X, Y, Z, levels=20, cmap="viridis")
fig.colorbar(cf, ax=axes[1])
axes[1].set_title("contourf")

fig.tight_layout()
plt.show()
```

## 9. Полярные координаты и параметрические кривые

```python
fig = plt.figure(figsize=(15, 4))

# Лиссажу — декартовы координаты, обычный subplot
ax1 = fig.add_subplot(1, 3, 1)
t = np.linspace(0, 2 * np.pi, 1000)
ax1.scatter(np.sin(3*t), np.sin(4*t), c=t, cmap="hsv", s=1)  # scatter + c=t для раскраски
ax1.set_title("Лиссажу")
ax1.set_aspect("equal")

# Спираль — полярные координаты: projection="polar"
ax2 = fig.add_subplot(1, 3, 2, projection="polar")
theta = np.linspace(0, 6*np.pi, 500)
ax2.plot(theta, 0 + 0.5 * theta)  # r = a + b*theta
ax2.set_title("Спираль Архимеда")

# Кардиоида — полярные координаты
ax3 = fig.add_subplot(1, 3, 3, projection="polar")
theta = np.linspace(0, 2*np.pi, 500)  # полный круг!
ax3.plot(theta, 1 + np.cos(theta))     # cos, не sin!
ax3.set_title("Кардиоида")

fig.tight_layout()
plt.show()
```

## Частые ошибки (показать студентам)

| Ошибка | Правильно |
|--------|----------|
| `arr.shape()` | `arr.shape` (атрибут, не метод) |
| `a * 2` (умножение) | `a ** 2` (возведение в степень) |
| `np.square(x)` ($x^2$) | `np.sqrt(x)` ($\sqrt{x}$) |
| `list + [10]` (конкатенация) | `np.array + 10` (broadcasting) |
| `plt.show()` перед `savefig()` | `savefig()` **перед** `show()` |
| `linewidth="2"` (строка) | `linewidth=2` (число) |
| `plt.style.use("seaborn")` | `plt.style.use("seaborn-v0_8")` |
| `np.sin(x) / x` (деление на 0) | `np.sinc(x / np.pi)` |
| `imshow(Z)` (перевёрнуто) | `imshow(Z, origin="lower")` |
| `projection="3d"` для полярных | `projection="polar"` |
