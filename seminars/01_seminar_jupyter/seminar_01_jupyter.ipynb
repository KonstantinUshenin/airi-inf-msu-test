### Бонус 3. Фрактал Мандельброта

**Найдите и исправьте ошибки (2 шт).** Код вычисляет множество Мандельброта, но:
- итерационная формула написана неправильно
- порог расходимости слишком маленький — картинка неинтересная

```python
# Сетка комплексных чисел
x = np.linspace(-2.5, 1.0, 800)
y = np.linspace(-1.25, 1.25, 800)
X, Y = np.meshgrid(x, y)
C = X + 1j * Y

Z = np.zeros_like(C)
mask = np.ones(C.shape, dtype=bool)
result = np.zeros(C.shape)

max_iter = 100
for i in range(max_iter):
    Z[mask] = Z[mask] ** 2 + Z[mask]  # итерация z = z^2 + c
    diverged = np.abs(Z) > 1.1
    result[mask & diverged] = i
    mask = mask & ~diverged

plt.figure(figsize=(10, 8))
plt.imshow(result, extent=[-2.5, 1.0, -1.25, 1.25], cmap="inferno", origin="lower")
plt.colorbar(label="Итерация расходимости")
plt.title("Множество Мандельброта")
plt.xlabel("Re")
plt.ylabel("Im")
plt.show()
```
